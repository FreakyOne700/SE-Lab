```python
# Non Refactored Code

def calc_stats(lst):
    # calculate average
    tot = 0
    for x in lst:
        tot += x
    avg = tot / len(lst)
    print("Avg: " + str(avg))
    
    lst2 = [10, 20, 30]
    tot2 = 0
    for y in lst2:
        tot2 += y
    avg2 = tot2 / len(lst2)
    print("Avg2: " + str(avg2))

calc_stats([2, 4, 6, 8])
```

```python
# Refactored Code

def calculate_average(numbers):
    """Extracted method to calculate the average of a list of numbers."""
    if not numbers:
        return 0
    total_sum = sum(numbers)
    return total_sum / len(numbers)

def display_statistics(primary_numbers, secondary_numbers):
    """Improved naming conventions and removed code duplication."""
    primary_average = calculate_average(primary_numbers)
    print(f"Primary Average: {primary_average}")
    
    secondary_average = calculate_average(secondary_numbers)
    print(f"Secondary Average: {secondary_average}")

# Execution
dataset_one = [2, 4, 6, 8]
dataset_two = [10, 20, 30]

display_statistics(dataset_one, dataset_two)
```

```python
class Task:
    def __init__(self, name, effort):
        self.name = name
        self.effort = effort  # Effort in hours or story points

    def __repr__(self):
        return f"{self.name} ({self.effort}h)"

class UserStory:
    def __init__(self, description):
        self.description = description
        self.tasks = []

    def add_task(self, task):
        self.tasks.append(task)

class Sprint:
    def __init__(self, capacity):
        self.capacity = capacity
        self.allocated_tasks = []
        self.current_effort = 0

    def allocate_task(self, task):
        """Allocates a task if sprint capacity allows."""
        if self.current_effort + task.effort <= self.capacity:
            self.allocated_tasks.append(task)
            self.current_effort += task.effort
            print(f"Allocated: {task.name} - Remaining Capacity: {self.capacity - self.current_effort}h")
        else:
            print(f"Skipped: {task.name} (Exceeds sprint capacity of {self.capacity}h)")

# --- Simulation Execution ---

# 1. Define a User Story
story = UserStory("As a user, I want to securely log into my account")

# 2. Break down user story into tasks and estimate effort
story.add_task(Task("Design login interface", 3))
story.add_task(Task("Setup database user table", 4))
story.add_task(Task("Implement JWT authentication logic", 8))
story.add_task(Task("Write unit tests for login", 4))

# 3. Create a Sprint and allocate tasks
sprint_1 = Sprint(capacity=16)

print(f"Starting Sprint Allocation (Total Capacity: {sprint_1.capacity}h)\n")
for task in story.tasks:
    sprint_1.allocate_task(task)

print(f"\nFinal Sprint Effort: {sprint_1.current_effort}h")
print(f"Tasks in Sprint: {sprint_1.allocated_tasks}")
```

```python
import time

class PairProgrammingSession:
    def __init__(self, student1, student2, switch_interval_minutes):
        self.driver = student1
        self.navigator = student2
        self.switch_interval = switch_interval_minutes

    def display_roles(self):
        print(f"\n--- Current Roles ---")
        print(f"Driver (Typing): {self.driver}")
        print(f"Navigator (Reviewing & Guiding): {self.navigator}")
        print("-" * 21)

    def switch_roles(self):
        """Switches the driver and navigator roles."""
        self.driver, self.navigator = self.navigator, self.driver
        print("\n*** TIME TO SWITCH ROLES! ***")

    def run_session(self, total_rounds):
        print("Starting Pair Programming Session...")
        for current_round in range(1, total_rounds + 1):
            print(f"\nRound {current_round} of {total_rounds}")
            self.display_roles()
            
            # Simulating the work period (in a real scenario, use time.sleep(self.switch_interval * 60))
            print(f"[Working for {self.switch_interval} minutes...]")
            time.sleep(1) # Using 1 second for simulation purposes
            
            if current_round < total_rounds:
                self.switch_roles()
                
        print("\nSession Complete! Great teamwork.")

# --- Execution ---
student_a = "Alice"
student_b = "Bob"
interval = 15 # minutes
rounds = 4

session = PairProgrammingSession(student_a, student_b, switch_interval_minutes=interval)
session.run_session(total_rounds=rounds)
```

```python
import unittest
from unittest.mock import MagicMock

# 1. The Dependency (e.g., an external API or Database)
class PaymentGateway:
    def charge(self, amount):
        # In a real scenario, this would make an external network request
        raise ConnectionError("Real payment gateway is not connected!")

# 2. The Component to Test (utilizing Dependency Injection)
class OrderProcessor:
    def __init__(self, payment_gateway):
        # The dependency is injected here, allowing us to swap it out later
        self.payment_gateway = payment_gateway 

    def checkout(self, amount):
        status = self.payment_gateway.charge(amount)
        return f"Checkout status: {status}"

# 3. The Unit Test (utilizing Mocking)
class TestOrderProcessor(unittest.TestCase):
    def test_checkout_success(self):
        # Create a mock object to replace the real PaymentGateway dependency
        mock_gateway = MagicMock()
        
        # Define what the mock should return when called
        mock_gateway.charge.return_value = "Success"

        # Inject the mock into our isolated component
        processor = OrderProcessor(mock_gateway)
        result = processor.checkout(100)

        # Assert the behavior is correct
        self.assertEqual(result, "Checkout status: Success")
        mock_gateway.charge.assert_called_once_with(100)
        
        print("Test Passed: The OrderProcessor was successfully isolated and tested using a Mock!")

# --- Execution ---
if __name__ == '__main__':
    # Running the test within the script
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

```python
import re

class SimulatedCodeAnalyzer:
    """Simulates a code quality analysis tool like CodeFactor or Codacy."""
    
    def __init__(self, code_snippet):
        self.code = code_snippet
        self.feedback = []
        self.issues_found = 0

    def analyze_style(self):
        """Checks for basic styling issues."""
        lines = self.code.split('\n')
        for i, line in enumerate(lines):
            if len(line) > 79:
                self.feedback.append(f"Line {i+1}: Style Warning - Line exceeds 79 characters.")
                self.issues_found += 1

    def analyze_complexity(self):
        """Simulates checking cyclomatic complexity."""
        if self.code.count("if ") + self.code.count("for ") + self.code.count("while ") > 3:
            self.feedback.append("Complexity Warning - High cyclomatic complexity detected. Consider refactoring.")
            self.issues_found += 1

    def analyze_potential_issues(self):
        """Checks for missing docstrings or missing type hints."""
        if "def " in self.code and '"""' not in self.code:
            self.feedback.append("Issue - Function missing docstring documentation.")
            self.issues_found += 1

    def run_review(self):
        print("🚀 Triggering Automated Code Review (GitHub Actions / CodeFactor Simulation)...\n")
        self.analyze_style()
        self.analyze_complexity()
        self.analyze_potential_issues()
        
        if self.issues_found == 0:
            print("✅ Status: PASS. Code looks great!")
        else:
            print(f"❌ Status: FAIL. Found {self.issues_found} issues to address:")
            for note in self.feedback:
                print(f"  -> {note}")

# --- Execution ---
if __name__ == "__main__":
    # Simulated code commit submitted in a Pull Request
    submitted_code = """
def process_data(data):
    for item in data:
        if item > 10:
            for x in range(item):
                if x % 2 == 0:
                    print("This is a very long line of code that definitely exceeds the standard PEP-8 seventy-nine character limit for Python styling.")
    return True
"""
    
    analyzer = SimulatedCodeAnalyzer(submitted_code)
    analyzer.run_review()
```
